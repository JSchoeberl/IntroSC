```python
import matplotlib.pyplot as plt
import numpy as np

n = 100
x = np.zeros(n)

x[0:20] = np.linspace(0.6,1,20)
x[20:70] = np.linspace(1,0.2,50)
x[70:100] = np.linspace(0.2,0.6,30)
```

```python
# Plot the values
plt.plot(x, '.')
plt.xlabel('Index')
plt.ylabel('Value')
plt.title(f'bitonic sequence')
# plt.grid(True)
plt.savefig('bitonic1.png')
plt.show()
```

```python
x2 = np.zeros(n)
x2[0:n//2] = np.minimum(x[0:n//2], x[n//2:n])
x2[50:100] = np.maximum(x[0:50], x[50:100])

x3 = np.zeros(n)
x3[0:n//4] = np.minimum(x2[0:n//4], x2[n//4:n//2])
x3[n//4:n//2] = np.maximum(x2[0:n//4], x2[n//4:n//2])
x3[n//2:3*n//4] = np.minimum(x2[n//2:3*n//4], x2[3*n//4:n])
x3[3*n//4:n] = np.maximum(x2[n//2:3*n//4], x2[3*n//4:n])
```

```python
plt.plot(x2, '.')
plt.xlabel('Index')
plt.ylabel('Value')
plt.title('bitonic merge')
# plt.grid(True)
plt.savefig('bitonic2.png')

plt.show()
```

```python
plt.plot(x3, '.')
plt.xlabel('Index')
plt.ylabel('Value')
plt.title('bitonic merge twice')
# plt.grid(True)
plt.savefig('bitonic3.png')
plt.show()
```
