# Various methods for the heat equation

Solve the heat equation from the previous notebook with various methods. Try also initial condition
$$
u_0 = \chi_{[0.4,0.6]^2}
$$
as a characteristic function.

We state the methods for the ODE 
$$
y^\prime(t) = f(t, y(t)),
$$
which is in our case
$$
y^\prime(t) = - M^{-1} A y(t)
$$

## Forward Euler method
$$
\frac{y_{n+1}-y_n}{\tau} = f(y_n)
$$

This method is conditional stable. Find timesteps $\tau$ such that it is working. How does $\tau$ depend on the mesh-size and the polynomial order?  

The so called CFL-condition gives a condition onto the timestep depending on $\rho (M^{-1} A)$

## Trapezoidal method = Crank Nicolson method

$$
\frac{y_{n+1} - y_n}{\tau} = \tfrac{1}{2} \left( f(y_n) + f(y_{n+1}) \right)
$$

* You should observe $2^{nd}$-order convergence in $\tau$
* Look at $u(x, 0.5)$ for small time, with discontinuous initial condition. There is the term Crank-Nicolson bump.

## DIRK methods

DIRK methods are a sub-class of RK methods, for which the $a$-matrix of the Butcher tableau is a lower triangular matrix:

$$
\begin{array}{c|cccc}
c_0 & a_{0,0} & 0 & \dots & 0 \\
\vdots & \vdots & \ddots & & \vdots \\
c_{s-1} & a_{s-1,0} & \dots & & a_{s-1,s-1} \\
\hline
 & b_0 & \dots & b_{s-1}
\end{array}
$$

An example is the third-order method

$$
\begin{array}{c|cccc}
c_0 & c_0 & 0 \\
1-c_0 & 1-2 c_0 & c_0 \\
\hline
 & 1/2 & 1/2 
\end{array}
\qquad \text{with} \qquad 
c_0 = \frac{3-\sqrt{3}}{6}
$$

with $f(u) = - M^{-1} A u$.

The Runge-Kutta formulas are

\begin{eqnarray*}
u^1 & = & u_n - \tau a_{11} M^{-1} A u^1 \\
u^2 & = & u_n - \tau a_{21} M^{-1} A u^1 - \tau a_{22} M^{-1} A u^2 \\
u_{n+1} & = & u_n - \frac{\tau}{2} M^{-1} A (u^1 + u^2)
\end{eqnarray*}

With the increments $w_1 = u^1 - u_n$ and $w_2 = u^2-u_n$ we obtain

\begin{eqnarray*}
(M + \tau a_{11} A) w_1 & = & -\tau a_{11} A u_n \\
(M + \tau a_{22} A) w_2 & = & -\tau a_{22} A u_n -\tau a_{21} A u^1 \\
\end{eqnarray*} 

and from
\begin{eqnarray*}
- M^{-1} A u_1 & = & \frac{1}{\tau a_{11}} (u^1 - u_n) = \frac{1}{\tau a_{11}} w_1 \\
- M^{-1} A u_2 & = & \frac{1}{\tau a_{22}} (w_2 - \frac{a_{21}}{a_{11}} w_1)
\end{eqnarray*}

we get
$$
u_{n+1} = u_n + \frac{1}{2} \left( \frac{1}{a_{11}} - \frac{a_{21}}{a_{11} a_{22}} \right) w_1  + \frac{1}{2 a_{22}} w_2
$$

## The stability function

* By using an eigenvector-basis (of $A z = \lambda Mz$) we can decompose the ODE $u^\prime = - M^{-1} A u$
  to scalar ODEs for $u_i^\prime = \lambda_i u_i$, with $\lambda_i \in {\mathbb R}^-$
* The stability function $g(z)$ is defined such that the numerical method applied to $y^\prime = \lambda y$ leads to 
    $$
     y_{n+1} = g(\tau \lambda) y_n
    $$

Examples for stability functions are:
* Forward Euler method:
  $$
  g(z) = 1+z
  $$
* Backward Euler method:
  $$
  g(z) = \frac{1}{1-z}
  $$
* Crank-Nicolson method:
  $$
  g(z) = \frac{2+z}{2-z}
  $$

The method is stable if $\forall \lambda \in \sigma(-M^{-1} A) : | g(\tau \lambda)| \leq 1$. 

For parabolic equations, $\sigma(-M^{-1} A) \subset {\mathbb R}^-$. A method is called {A-stable} if 
$\forall \lambda \in {\mathbb R}^- : | g(\tau \lambda)| \leq 1$. 

For large negative $\lambda$, the solution $u_i u_{i,0} e^{-\lambda_i t}$ is falling quickly. A method is called {\em L-stable} if it is A-stable, and 
$$
\lim_{z \rightarrow \infty} g(z) = 0
$$
